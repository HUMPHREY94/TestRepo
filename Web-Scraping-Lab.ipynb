## Extract information from the given web site
You will extract the data from the below web site: <br> 


```python
#this url contains the data you need to scrape
url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DA0321EN-SkillsNetwork/labs/datasets/Programming_Languages.html"
```

The data you need to scrape is the **name of the programming language** and **average annual salary**.<br> It is a good idea to open the url in your web broswer and study the contents of the web page before you start to scrape.


Import the required libraries


```python
# Your code here
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

Download the webpage at the url


```python
#your code goes here
import requests



url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DA0321EN-SkillsNetwork/labs/datasets/Programming_Languages.html"


response = requests.get(url)


if response.status_code == 200:
    print("Webpage downloaded successfully!")
    html_content = response.text  
else:
    print(f"Failed to retrieve the webpage. Status code: {response.status_code}")
```

Create a soup object


```python
#your code goes here

from bs4 import BeautifulSoup


soup = BeautifulSoup(html_content, 'html.parser')

```

Scrape the `Language name` and `annual average salary`.


```python
#your code goes here

tables = soup.find_all('table')


for i, table in enumerate(tables):
    print(f"\nTable {i} preview:")
    for row in table.find_all('tr')[:2]:  
        print([col.text.strip() for col in row.find_all(['td', 'th'])])


table = tables[0]  
languages = []
salaries = []


for row in table.find_all('tr')[1:]:
    cols = row.find_all(['td', 'th'])  
    if len(cols) >= 2:
        language = cols[0].get_text(strip=True)
        salary = cols[1].get_text(strip=True)
        languages.append(language)
        salaries.append(salary)


print("\nLanguage : Average Salary")
for lang, sal in zip(languages, salaries):
    print(f"{lang} : {sal}")
```

Save the scrapped data into a file named *popular-languages.csv*


```python
# your code goes here
import csv


filename = 'popular-languages.csv'


with open(filename, mode='w', newline='', encoding='utf-8') as file:
    writer = csv.writer(file)
    
    writer.writerow(['Language', 'Average Salary'])
    
    for lang, sal in zip(languages, salaries):
        writer.writerow([lang, sal])

print(f"Data saved to {filename}")
```
